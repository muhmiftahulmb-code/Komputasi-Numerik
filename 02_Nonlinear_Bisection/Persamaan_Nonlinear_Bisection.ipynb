## 1. Dasar teori

Metode Bisection mencari pendekatan akar persamaan nonlinear satu variabel `f(x) = 0`. Pertama, dipilih interval `[a, b]` yang memenuhi `f(a) × f(b) < 0`. Jika fungsi kontinu pada interval tersebut, perubahan tanda menunjukkan bahwa interval mengurung setidaknya satu akar.

Titik tengah dihitung dengan rumus `c = (a + b) / 2`. Selanjutnya dipilih setengah interval yang tetap memiliki perubahan tanda. Proses diulang hingga galat memenuhi toleransi atau iterasi mencapai batas maksimum.

## 2. Tentukan fungsi dan parameter

Contoh yang digunakan adalah `f(x) = x^3 - 4x - 9` dengan interval awal `[2, 3]`. Nilai fungsi pada kedua ujung interval memiliki tanda berbeda, sehingga interval memenuhi syarat awal Bisection. Toleransi menentukan ketelitian pendekatan akar.

```python
# Definisikan fungsi nonlinear yang akan dicari akarnya.
def f(x):
    return x**3 - 4*x - 9

# Tentukan batas bawah dan batas atas interval pencarian.
a = 2.0
b = 3.0

# Tentukan batas galat dan jumlah iterasi maksimum.
toleransi = 1e-6
iterasi_maksimum = 100

# Tampilkan nilai fungsi di kedua ujung interval.
print(f"f(a) = f({a}) = {f(a):.6f}")
print(f"f(b) = f({b}) = {f(b):.6f}")
```

## 3. Membuat fungsi Bisection

Fungsi berikut memeriksa nilai batas, memastikan interval mengurung akar, menghitung titik tengah, dan memperbarui interval berdasarkan tanda fungsi. Riwayat iterasi disimpan agar prosesnya dapat dianalisis.

```python
def bisection(fungsi, batas_bawah, batas_atas, toleransi=1e-6, maksimum=100):
    # Pastikan toleransi bernilai positif dan jumlah iterasi masuk akal.
    if toleransi <= 0 or maksimum < 1:
        raise ValueError("Toleransi harus positif dan iterasi maksimum minimal 1.")

    # Hitung nilai fungsi pada kedua ujung interval.
    fa = fungsi(batas_bawah)
    fb = fungsi(batas_atas)

    # Jika batas bawah atau batas atas sudah merupakan akar, langsung kembalikan.
    if fa == 0:
        return batas_bawah, [], True
    if fb == 0:
        return batas_atas, [], True

    # Bisection memerlukan perubahan tanda pada interval awal.
    if fa * fb > 0:
        raise ValueError("f(a) dan f(b) harus berlainan tanda.")

    # Siapkan daftar untuk menyimpan hasil setiap iterasi.
    riwayat = []
    konvergen = False
    titik_tengah = None

    # Lakukan pembagian interval sampai memenuhi kriteria berhenti.
    for iterasi in range(1, maksimum + 1):
        # Hitung titik tengah interval saat ini.
        titik_tengah = (batas_bawah + batas_atas) / 2
        fc = fungsi(titik_tengah)

        # Simpan informasi iterasi sebelum interval diperbarui.
        riwayat.append((
            iterasi, batas_bawah, batas_atas, titik_tengah, fc,
            (batas_atas - batas_bawah) / 2
        ))

        # Berhenti jika nilai fungsi atau setengah lebar interval cukup kecil.
        if abs(fc) <= toleransi or (batas_atas - batas_bawah) / 2 <= toleransi:
            konvergen = True
            break

        # Pertahankan setengah interval yang memiliki perubahan tanda.
        if fa * fc < 0:
            batas_atas = titik_tengah
            fb = fc
        else:
            batas_bawah = titik_tengah
            fa = fc

    # Kembalikan pendekatan akar, tabel iterasi, dan status konvergensi.
    return titik_tengah, riwayat, konvergen
```

## 4. Jalankan metode Bisection

Cell ini menjalankan fungsi menggunakan parameter yang telah ditentukan. Tabel menampilkan batas interval, titik tengah, nilai fungsi, dan setengah lebar interval pada setiap iterasi.

```python
# Jalankan Bisection dan simpan hasil serta riwayat perhitungan.
akar, riwayat, konvergen = bisection(
    f, a, b, toleransi=toleransi, maksimum=iterasi_maksimum
)

# Tampilkan tabel ringkas seluruh iterasi.
print("Iter |       a       |       b       |       c       |      f(c)      | Setengah lebar")
print("-" * 89)
for baris in riwayat:
    n, a_i, b_i, c_i, fc_i, setengah_lebar = baris
    print(
        f"{n:4d} | {a_i:13.8f} | {b_i:13.8f} | "
        f"{c_i:13.8f} | {fc_i:13.6e} | {setengah_lebar:13.6e}"
    )

# Tampilkan hasil akhir dan status iterasi.
print("\nPendekatan akar:", f"{akar:.10f}")
print("Nilai fungsi di pendekatan akar:", f"{f(akar):.10e}")
print("Jumlah iterasi:", len(riwayat))
print("Status:", "Konvergen" if konvergen else "Batas iterasi tercapai")
```

## 5. Verifikasi hasil

Akar pendekatan diperiksa kembali dengan menghitung `f(akar)`. Nilai yang mendekati nol menunjukkan bahwa hasil pendekatan memenuhi persamaan dengan galat kecil. Besar toleransi dan interval awal memengaruhi proses serta ketelitian hasil.

```python
# Hitung nilai fungsi pada pendekatan akar yang diperoleh.
nilai_pemeriksaan = f(akar)

# Bandingkan nilai fungsi dengan toleransi yang ditentukan.
print(f"Akar pendekatan = {akar:.10f}")
print(f"f(akar) = {nilai_pemeriksaan:.10e}")

if abs(nilai_pemeriksaan) <= toleransi or konvergen:
    print("Pemeriksaan selesai: pendekatan akar memenuhi kriteria penghentian.")
else:
    print("Pendekatan belum memenuhi kriteria; pertimbangkan menambah iterasi.")
```

Metode Bisection mempersempit interval yang mengurung akar dengan cara membagi interval menjadi dua bagian pada setiap iterasi. Metode ini sederhana dan memiliki proses yang terkontrol selama fungsi kontinu dan syarat perubahan tanda dipenuhi.
