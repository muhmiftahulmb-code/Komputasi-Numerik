## 1. Dasar teori

SPL tiga variabel terdiri dari tiga persamaan linear yang memuat variabel `x`, `y`, dan `z`. Metode Gauss-Jordan mengubah matriks augmented menjadi bentuk eselon baris tereduksi sehingga nilai variabel dapat dibaca pada kolom konstanta.

Operasi yang digunakan adalah pertukaran baris, perkalian satu baris dengan konstanta bukan nol, dan penambahan kelipatan suatu baris ke baris lain. Program memilih pivot dengan nilai mutlak terbesar pada kolom aktif (pivot parsial) untuk mengurangi risiko menggunakan pivot yang sangat kecil.

## 2. Masukkan koefisien persamaan

Contoh sistem yang digunakan:

- `2x + y - z = 8`
- `-3x - y + 2z = -11`
- `-2x + y + 2z = -3`

Matriks `A` menyimpan koefisien variabel, sedangkan `b` menyimpan konstanta di ruas kanan. Kamu dapat mengganti nilai ini dengan soal dari dosen.

```python
# Matriks koefisien dari tiga persamaan (kolom berurutan: x, y, z).
A = [
    [ 2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
]

# Vektor konstanta dari ruas kanan setiap persamaan.
b = [8, -11, -3]

# Tampilkan data awal agar mudah diperiksa.
print("Matriks koefisien A:")
for baris in A:
    print(baris)
print("Konstanta b:", b)
```

## 3. Membuat fungsi Gauss-Jordan

Fungsi berikut melakukan proses eliminasi. Setiap baris matriks augmented diproses untuk mendapatkan pivot 1, lalu elemen lain pada kolom pivot dibuat menjadi 0. Pemilihan pivot parsial dilakukan sebelum normalisasi baris.

```python
def gauss_jordan_3x3(A, b, toleransi=1e-12):
    # Pastikan matriks mempunyai tiga baris dan setiap baris memiliki tiga koefisien.
    if len(A) != 3 or any(len(baris) != 3 for baris in A):
        raise ValueError("Matriks koefisien harus berukuran 3 x 3.")

    # Pastikan vektor konstanta mempunyai tiga elemen.
    if len(b) != 3:
        raise ValueError("Vektor konstanta harus mempunyai tiga elemen.")

    # Gabungkan A dan b menjadi matriks augmented [A | b].
    matriks = [
        [float(nilai) for nilai in A[i]] + [float(b[i])]
        for i in range(3)
    ]

    # Proses setiap kolom variabel sebagai kolom pivot.
    for kolom in range(3):
        # Pilih baris dengan nilai absolut terbesar pada kolom aktif.
        baris_pivot = max(
            range(kolom, 3),
            key=lambda i: abs(matriks[i][kolom])
        )

        # Tukar baris aktif dengan baris yang mempunyai pivot terbaik.
        matriks[kolom], matriks[baris_pivot] = matriks[baris_pivot], matriks[kolom]

        # Ambil nilai pivot setelah pertukaran baris.
        pivot = matriks[kolom][kolom]

        # Hentikan jika pivot terlalu kecil; solusi tunggal tidak dapat dipastikan.
        if abs(pivot) < toleransi:
            raise ValueError("Matriks singular atau tidak memiliki solusi tunggal.")

        # Bagi seluruh elemen baris pivot agar nilai pivot menjadi 1.
        for j in range(4):
            matriks[kolom][j] /= pivot

        # Hilangkan elemen pada kolom pivot dari dua baris lainnya.
        for i in range(3):
            if i != kolom:
                faktor = matriks[i][kolom]

                # Kurangkan kelipatan baris pivot dari baris target.
                for j in range(4):
                    matriks[i][j] -= faktor * matriks[kolom][j]

    # Ambil kolom terakhir sebagai nilai x, y, dan z.
    solusi = [matriks[i][3] for i in range(3)]

    # Kembalikan solusi dan matriks akhir untuk ditampilkan.
    return solusi, matriks
```

## 4. Jalankan proses penyelesaian

Cell ini memanggil fungsi yang sudah dibuat. Hasil yang diharapkan untuk contoh data adalah `x = 2`, `y = 3`, dan `z = -1`.

```python
# Jalankan metode Gauss-Jordan pada data A dan b.
solusi, matriks_akhir = gauss_jordan_3x3(A, b)

# Tampilkan bentuk matriks setelah proses eliminasi.
print("Matriks setelah eliminasi Gauss-Jordan:")
for baris in matriks_akhir:
    print([round(nilai, 10) for nilai in baris])

# Tampilkan nilai setiap variabel.
x, y, z = solusi
print("\nHasil penyelesaian:")
print(f"x = {x:.6f}")
print(f"y = {y:.6f}")
print(f"z = {z:.6f}")
```

## 5. Verifikasi hasil

Pemeriksaan dilakukan dengan memasukkan kembali nilai `x`, `y`, dan `z` ke setiap persamaan awal. Nilai ruas kiri seharusnya mendekati konstanta pada ruas kanan. Selisih kecil dapat muncul karena representasi bilangan pecahan oleh komputer.

```python
# Hitung ulang ruas kiri dari setiap persamaan menggunakan solusi.
print("Pemeriksaan ke persamaan awal:")
for i in range(3):
    ruas_kiri = sum(A[i][j] * solusi[j] for j in range(3))
    selisih = ruas_kiri - b[i]
    print(
        f"Persamaan {i + 1}: ruas kiri = {ruas_kiri:.10f}, "
        f"ruas kanan = {b[i]:.10f}, selisih = {selisih:.2e}"
    )

# Kesimpulan verifikasi berdasarkan batas galat numerik.
if all(abs(sum(A[i][j] * solusi[j] for j in range(3)) - b[i]) < 1e-9 for i in range(3)):
    print("Verifikasi berhasil: solusi memenuhi ketiga persamaan.")
else:
    print("Verifikasi belum memenuhi toleransi yang ditetapkan.")
```
